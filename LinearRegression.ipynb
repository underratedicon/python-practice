```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn import datasets, model_selection
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
df = pd.read_csv('sample_data/BostonHousing.csv')
df.head()
```

```python
df = df.rename(columns={
    'medv': 'TARGET',
    'rm': 'RM',
    'lstat': 'LSTAT'
})
```

```python
df.head()
```

```python
df[["LSTAT", "RM", "TARGET"]].head()
```

```python
plt.figure(figsize=(10,5))

features = df[["LSTAT", "RM"]]
target = df["TARGET"]
```

```python
plt.figure(figsize=(8,4))

features = df[["LSTAT", "RM"]]
target = df["TARGET"]

plt.subplot(1,2,1)
plt.scatter(df["LSTAT"], target)
plt.xlabel('LSTAT')
plt.ylabel('target')

plt.subplot(1,2,2)
plt.scatter(df["RM"], target)
plt.xlabel('LSTAT')
plt.ylabel('target')

plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=23)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
plt.figure(figsize=(8,4))
plt.scatter(y_test, y_pred)
plt.plot([0, 50], [0, 50], '--k', label="Correct prediction")
plt.xlabel('Actual y')
plt.ylabel('Predicted y')
plt.show()
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
intercept = model.intercept_
coff = model.coef_
```

```python
# Evaluate the model's performance
mse = mean_squared_error(y_test, y_pred)
print('Mean Square Error :', mse)

mae = mean_absolute_error(y_test, y_pred)
print('Mean Absolute Error:', mae)

r2 = r2_score(y_test, y_pred)
print('R2            :', r2)
```
